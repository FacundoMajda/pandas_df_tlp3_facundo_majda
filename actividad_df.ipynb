```python
import pandas as pd
import sqlite3

db_path = 'Salaries.sqlite'

connection = sqlite3.connect(db_path)

query = '''SELECT * FROM Salaries'''

df = pd.read_sql_query(query, connection)
connection.close()

print(df.head())
print(df.info())
```

1. **Top 10 empleados con mayor salario total (incluyendo beneficios)**
   Extrae e imprime un top 10 de expleados con mayor salario incluyendo los beneficios


```python
top_10=df.nlargest(10, 'TotalPayBenefits')

print(top_10[['EmployeeName', 'TotalPayBenefits']])
```

2. **Filtrar empleados con más de 50,000 en horas extra, la columna a tratar es: (OvertimePay)**


```python
df['OvertimePay'] = pd.to_numeric(df['OvertimePay'], errors='coerce')

overtime = df[df['OvertimePay'] > 50000]
print(overtime[['EmployeeName', 'OvertimePay']])

```

3. **Contar cuántos empleados únicos hay por año**


```python
unique_employees_per_year = df.groupby('Year')['EmployeeName'].nunique()
print(unique_employees_per_year)
```

4. **Ver cuántos cargos únicos (JobTitle) existen y los 5 más comunes**


```python
num_unique_jobs = df['JobTitle'].nunique()
print(f"Cantidad de cargos unicos: {num_unique_jobs}")

top_5 = df['JobTitle'].value_counts().head(5)
print("Los 5 mas comunes:")
print(top_5)
```

5. **Mostrar el salario total (TotalPay) promedio por año**


```python
average_total_pay_per_year = df.groupby('Year')['TotalPay'].mean()


print("(TotalPay) promedio por año:")
print(average_total_pay_per_year)
```

6. **Exporta todos los datos a un CSV y muestra los ultimos 10 valores**


```python
# el nombre del archivo CSV
csv_file_path = 'salaries_exported.csv'

# exportar el df a csv sin incluir el índice de pandas
df.to_csv(csv_file_path, index=False)
print(f"datos exportados a {csv_file_path}")

# mostrar los últimos 10 registros
db = pd.read_csv(csv_file_path)
print("ultimos 10 registros:")
print(db.tail(10))
```
